can u delete individual element in tuple ,in list,in set,in



dict
Here is the short answer for all four:

Tuple: No. (Tuples are immutable/unchangeable. You can't delete items, only delete the whole tuple).

List: Yes. Use del my_list[index] or my_list.remove(value).

Set: Yes. Use my_set.remove(value) or my_set.discard(value).

Dict: Yes. Use del my_dict[key] or my_dict.pop(key).
