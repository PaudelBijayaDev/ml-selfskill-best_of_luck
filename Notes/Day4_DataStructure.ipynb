Interview q3: deep copy and shallow copy

```python
#shallow copy
lst1=[[1,2,3],[4,5,6]]
lst2=lst1.copy()#copied original above lis1 to lst2
lst1[1][1] = 100 #change made to lst2
print(lst1)
print(lst2) #but changed seen on lst2 but only 1 was changed
```

```python
#deep copy
import copy
lst3=[[1,2,3],[4,5,6]] #simply made new list
lst4 = copy.deepcopy(lst3) #here special deepcopy to 4
lst3[1][2] = 100 #change made to lst3,also this is 2d list
print(lst3) #change seen only on original lst3
print(lst4)
#if single or 1d list then no change in shallow or deep copy
```
