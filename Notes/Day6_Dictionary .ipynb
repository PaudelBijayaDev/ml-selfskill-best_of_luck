```python
#examples

#create dictionary using {}
man = {"name":"Bijaya", "age":28,"country":"Nepal", "contact":9860914199}
print(man)

#create dictionary using dict()
gentleman = dict({"name":"Bijaya", "age":28,"country":"Nepal", "contact":9860914199})
print(gentleman) #diff-()

#create dict from sequence
gentleman = dict([("name","Bijaya"),("age",28),("country","Nepal"),("contact",9860914199)])
print(gentleman)#diff-[],comma instead of :,()

#create dict with mixed keys
sample = {"name":"bijaya",11:"Nepal"}
print(sample)#key is string and int

#create dict using value as list-when values are multiple lik mob no
person = {"name":"bijaya","mobile":[9860914199,9844342893]}
print(person)
```

```python
#dictionary methods

man = {"name":"Bijaya", "age":28,"country":"Nepal", "contact":9860914199}

#displays all keys in dict
print(man.keys())


#displays all value in dict
print(man.values())

#return all items in dict inside tuple
print(man.items())

#pop(key[,d])

print(man.popitem())

del man['age']
dict(man)

# clear()-deletes all inside dict

#del man-deletes whole dictionary
```

```python
man = {"name":"Bijaya", "age":28,"country":"Nepal", "contact":9860914199}

# 1. remove last inserted item from dict
print(man.popitem())

# 2.display updated dict
print(man)

new = man.pop('age')
print(new)#removes key called age

```

```python
#join two dictionary


# using update method
dict1 = {'Bijaya':27,'Chris':29,'Smith':39}
dict2 = {'Rita':38,'Sita':39,'Alica':37}
print(dict1,", original dict1")
dict1.update(dict2)
print(dict1,", updted dict1")


# using **kwargs to unpack
kwarg_dict={**dict1,**dict2}
print(kwarg_dict,", used kwarg")
```

```python
#sorted method in dict
samp = {'c':43,'b':47}
print(sorted(samp.items()))
print(samp)
```

```python
#Dictionary Operations

# 1.make a dict using constructor-ie use dict,means must be ()outside,2bracs
dicn = dict({'name':'Bijaya','Passion':'Wealth Generation'})
dicn

# 2 create an empty dictionary
d1 = {}

# 3 retrieve value using keyname
d1.get('a')

# 4 returns list of keys in dict
d1.keys()

# 5 returs list with values
d1.values()

# 6 return with key value pair
d1.items()

# 7 return no of items in dict
len(d1)

# 8 update dic by adding key
d1['d'] = 40

# 9 add multiple keys to dict
d1.update({'a':34, 'b':50})

# 10 set default value if no key is there
d1.setdefault('c',30)

# 11 modify value of existing key
d1['b'] = 20

# 12 remove key b from dic
d1.pop('b')

# 13 remove random item frm dict
d1.popitem()

# 14 remove all item from dict
d1.clear()

# 15 check if key exists in dict
'key' in d1.keys()

# 16 add all item of d1 into d2
d1.update(d1)

# 17 join two dict
d3 = {**d1, **d1}

# 18 copy dict d1 into d2
d2 = d1.copy()

# 19 return key with max value
max(d1)#min(d1)


```
